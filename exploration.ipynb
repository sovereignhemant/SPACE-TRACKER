# India Space Policy & Regulatory Regime Tracker - Data Exploration

This notebook conducts analytical exploration on `policy_events.csv` and `regimes.csv`, deriving key metrics on policy velocity, authorization trends, and regime benchmarking.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# Load data
data_dir = Path('../data')
events_df = pd.read_csv(data_dir / 'policy_events.csv')
regimes_df = pd.read_csv(data_dir / 'regimes.csv')

events_df['date'] = pd.to_datetime(events_df['date'])
print(f"Total policy events loaded: {len(events_df)}")
events_df.head(5)
```

## 1. Distribution of Policy Events by Category & Country

```python
ct = pd.crosstab(events_df['country'], events_df['category'])
print(ct)

ct.plot(kind='bar', stacked=True, figsize=(10, 5))
plt.title('Space Policy Events by Country and Category')
plt.ylabel('Number of Events')
plt.xlabel('Regime')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 2. IN-SPACe Authorization & Launch Velocity (India)

```python
india_auths = events_df[(events_df['country'] == 'India') & (events_df['category'].isin(['authorization', 'launch_milestone']))].sort_values('date')
india_auths['cumulative'] = range(1, len(india_auths) + 1)

plt.figure(figsize=(10, 4))
plt.plot(india_auths['date'], india_auths['cumulative'], marker='o', color='green', linewidth=2)
plt.title('Cumulative IN-SPACe Authorizations & Launch Milestones (2020-2026)')
plt.xlabel('Date')
plt.ylabel('Cumulative Count')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

## 3. FDI Policy Implementation Lag Analysis

Calculating time lag between Cabinet policy approval and statutory FEMA notification in India.

```python
cab_date = events_df[events_df['event_id'] == 'fdi-policy-space-cabinet-2024']['date'].values[0]
fema_date = events_df[events_df['event_id'] == 'fdi-policy-fema-notification-2024']['date'].values[0]
lag_days = (pd.to_datetime(fema_date) - pd.to_datetime(cab_date)).days
print(f"FDI Policy Cabinet Approval Date: {pd.to_datetime(cab_date).strftime('%Y-%m-%d')}")
print(f"FEMA Notification Date: {pd.to_datetime(fema_date).strftime('%Y-%m-%d')}")
print(f"Policy Enactment Lag: {lag_days} days")
```
